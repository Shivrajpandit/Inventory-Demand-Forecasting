# Notebook 04: Machine Learning Model Training
## Comparing Ridge, Random Forest, LightGBM, and XGBoost

This notebook trains multiple machine learning regressors on the chronologically split training set and tunes hyperparameters.

```python
import os
import pandas as pd
import numpy as np

from ml.features.feature_pipeline import FeaturePipeline
from ml.models.linear_model import LinearDemandModel
from ml.models.tree_models import RandomForestDemandModel, XGBoostDemandModel
from ml.models.lightgbm_model import LightGBMDemandModel
from ml.evaluation.metrics import evaluate_forecast
```

```python
# Load features and split
df = pd.read_csv('../data/processed/cleaned_sales_data.csv')
pipeline = FeaturePipeline()
df_feat = pipeline.transform(df, drop_na=True)
train_df, val_df, test_df = FeaturePipeline.chronological_split(df_feat, train_ratio=0.70, val_ratio=0.15)

print(f'Train rows: {len(train_df):,} | Val rows: {len(val_df):,} | Test rows: {len(test_df):,}')
```

```python
# Train XGBoost Regressor
xgb_model = XGBoostDemandModel(n_estimators=150, max_depth=6, learning_rate=0.05)
xgb_model.fit(train_df, val_df=val_df)

val_preds = xgb_model.predict(val_df)
val_metrics = evaluate_forecast(val_df['Units_Sold'].values, val_preds)
print('XGBoost Validation Metrics:', val_metrics)
```
