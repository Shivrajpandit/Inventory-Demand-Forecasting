# Notebook 05: Model Evaluation & Feature Attribution
## Benchmark Comparison, Error Diagnostics & Explainability

This notebook evaluates all models on the out-of-time test split and visualizes feature importances and residuals.

```python
import os
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from ml.forecasting.registry import ModelRegistry

sns.set_theme(style='whitegrid')
```

```python
# Load model comparison results
with open('../docs/model_comparison.json', 'r') as f:
    results = json.load(f)

comp_df = pd.DataFrame(results).sort_values(by='MAE')
comp_df[['Model', 'MAE', 'RMSE', 'WAPE_pct', 'sMAPE_pct']]
```

```python
# Load Champion Model & Feature Importances
registry = ModelRegistry(registry_dir='../models')
meta = registry.load_champion_metadata()
fi = meta.get('feature_importance', {})

fi_df = pd.DataFrame(list(fi.items()), columns=['Feature', 'Importance']).head(12)

plt.figure(figsize=(10, 6))
sns.barplot(data=fi_df, x='Importance', y='Feature', palette='mako')
plt.title(f'Feature Importance: {meta["model_name"]}', fontsize=14, fontweight='bold')
plt.xlabel('Importance Score')
plt.ylabel('Feature')
plt.show()
```
