# Notebook 02: Exploratory Data Analysis (EDA)
## Demand Patterns, Seasonality, Promotions & Inventory Dynamics

This notebook analyzes:
- Aggregate sales trends and distributions
- Product & category demand velocity
- Day-of-week and monthly seasonality patterns
- Impact of price discounts and promotions
- Holiday demand surges
- Stockout frequency and inventory behavior

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from ml.preprocessing.eda import RetailEDA

# Set visual style
sns.set_theme(style='whitegrid', palette='deep')
plt.rcParams['figure.figsize'] = (12, 6)
```

```python
# Load cleaned sales data
df = pd.read_csv('../data/processed/cleaned_sales_data.csv')
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```

```python
# Run automated EDA engine and export report
eda = RetailEDA(output_dir='../docs/eda_reports')
metrics = eda.run_full_eda(df)
metrics['overall']
```

```python
# Day of Week Analysis
dow_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
dow_map = {i: dow_order[i] for i in range(7)}
df['Day_Name'] = df['Day_of_Week'].map(dow_map)

plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='Day_Name', y='Units_Sold', order=dow_order, palette='Blues_d')
plt.title('Average Daily Demand by Day of Week', fontsize=14, fontweight='bold')
plt.ylabel('Average Units Sold')
plt.show()
```

```python
# Promotional Uplift Analysis
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Promotion', y='Units_Sold', palette=['#94a3b8', '#10b981'], showfliers=False)
plt.xticks([0, 1], ['No Promo', 'Active Promo'])
plt.title('Demand Comparison: Standard vs Promotional Days', fontsize=14, fontweight='bold')
plt.ylabel('Daily Units Sold')
plt.show()
```
